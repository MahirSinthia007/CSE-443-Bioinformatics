```python
def prefix_function(p):
    m=len(p)
    pie={}
    pie[0]=0
    k=0
    for q in range(1,m):
        while k>0 and p[k]!=p[q]:
            k=pie[k-1]
        if p[k]==p[q]:
            k=k+1
        pie[q]=k
    return pie

```
